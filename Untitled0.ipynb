```python
import time
import requests
from bs4 import BeautifulSoup
from pydantic import BaseModel, Field
from typing import List, Optional

# 1. Standart İlan Veri Modeli (Pydantic)
class JobListing(BaseModel):
    title: str = Field(description="İş ilanı başlığı")
    company: str = Field(description="Şirket adı")
    location: str = Field(description="Konum")
    job_link: str = Field(description="İlanın doğrudan bağlantı linki")
    job_id: str = Field(description="İlanın benzersiz kimliği")

# 2. LinkedIn Açık İlan Çekme Fonksiyonu
def fetch_linkedin_jobs(keyword: str, location: str = "Turkey", limit: int = 5) -> List[JobListing]:
    """
    Belirtilen anahtar kelime ve konum için LinkedIn'den temel ilan listesini çeker.
    """
    # LinkedIn Guest API arama URL'i
    url = f"https://www.linkedin.com/jobs-guest/jobs/api/seeMoreJobPostings/search?keywords={keyword}&location={location}&start=0"

    # Bot engellerine takılmamak için tarayıcı kimliği (Header)
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "tr-TR,tr;q=0.9,en-US;q=0.8,en;q=0.7"
    }

    job_list = []

    try:
        response = requests.get(url, headers=headers, timeout=10)

        if response.status_code != 200:
            print(f"[-] İstek başarısız oldu. Durum Kodu: {response.status_code}")
            return []

        soup = BeautifulSoup(response.text, "html.parser")
        job_cards = soup.find_all("li")

        for card in job_cards:
            if len(job_list) >= limit:
                break

            title_tag = card.find("h3", class_="base-search-card__title")
            company_tag = card.find("h4", class_="base-search-card__subtitle")
            location_tag = card.find("span", class_="job-search-card__location")
            link_tag = card.find("a", class_="base-card__full-link")

            if title_tag and company_tag and link_tag:
                title = title_tag.get_text(strip=True)
                company = company_tag.get_text(strip=True)
                loc = location_tag.get_text(strip=True) if location_tag else "Belirtilmemiş"
                link = link_tag.get("href", "").split("?")[0] # Linki temizle

                # Link içerisinden Job ID ayıklama
                job_id = link.rstrip("/").split("-")[-1]

                job = JobListing(
                    title=title,
                    company=company,
                    location=loc,
                    job_link=link,
                    job_id=job_id
                )
                job_list.append(job)

        print(f"[+] '{keyword}' araması için {len(job_list)} adet ilan başarıyla çekildi.")
        return job_list

    except Exception as e:
        print(f"[-] Hata oluştu: {str(e)}")
        return []

# --- TEST ÇALIŞTIRMASI ---
if __name__ == "__main__":
    sample_jobs = fetch_linkedin_jobs(keyword="Data Analyst", location="Istanbul, Turkey", limit=3)

    print("\n--- ÇEKİLEN ÖRNEK İLANLAR ---")
    for idx, job in enumerate(sample_jobs, 1):
        print(f"{idx}. {job.title} | {job.company} ({job.location})")
        print(f"   Link: {job.job_link}\n")
```

```python
import time
import json
import requests
from bs4 import BeautifulSoup
from typing import List, Optional
from pydantic import BaseModel, Field
from google import genai
from google.genai import types
from google.colab import userdata

# ==========================================
# 1. GEMINI API İSTEMCİSİ BAŞLATMA
# ==========================================
try:
    api_key = userdata.get('GEMINI_API_KEY')
except Exception:
    api_key = "AQ.Ab8RN6JYl_qsomHqOu4uHLACl-PQgspbptq_fxJxKsly00LHJg"

client = genai.Client(api_key=api_key)

# ==========================================
# 2. VERİ MODELLERİ (PYDANTIC)
# ==========================================
class JobListing(BaseModel):
    title: str = Field(description="İş ilanı başlığı")
    company: str = Field(description="Şirket adı")
    location: str = Field(description="Konum")
    job_link: str = Field(description="İlanın doğrudan bağlantı linki")
    job_id: str = Field(description="İlanın benzersiz kimliği")

class JobMatchAnalysis(BaseModel):
    match_score: int = Field(description="Profil ile ilan arasındaki 100 üzerinden uyum puanı (0-100)")
    suitability_category: str = Field(description="'Yüksek Uyum', 'Orta Uyum' veya 'Düşük Uyum'")
    extracted_requirements: List[str] = Field(description="İlandan ayıklanan temel teknik/sosyal beklentiler ve aranan nitelikler")
    matching_points: List[str] = Field(description="Adayın profiliyle örtüşen güçlü yönler")
    missing_or_risk_points: List[str] = Field(description="Adayda eksik olabilecek veya ilanın getirdiği riskler/yüksek tecrübe şartı")
    brief_summary: str = Field(description="İlanın odak noktası ve aday için 2 cümlelik net değerlendirme")

# ==========================================
# 3. İLAN ÇEKME VE DETAY FONKSİYONLARI (FAZ 1)
# ==========================================
def fetch_linkedin_jobs(keyword: str, location: str = "Turkey", limit: int = 3) -> List[JobListing]:
    """LinkedIn'den temel ilan listesini çeker."""
    url = f"https://www.linkedin.com/jobs-guest/jobs/api/seeMoreJobPostings/search?keywords={keyword}&location={location}&start=0"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "tr-TR,tr;q=0.9,en-US;q=0.8,en;q=0.7"
    }

    job_list = []
    try:
        response = requests.get(url, headers=headers, timeout=10)
        if response.status_code != 200:
            return []

        soup = BeautifulSoup(response.text, "html.parser")
        job_cards = soup.find_all("li")

        for card in job_cards:
            if len(job_list) >= limit:
                break

            title_tag = card.find("h3", class_="base-search-card__title")
            company_tag = card.find("h4", class_="base-search-card__subtitle")
            location_tag = card.find("span", class_="job-search-card__location")
            link_tag = card.find("a", class_="base-card__full-link")

            if title_tag and company_tag and link_tag:
                title = title_tag.get_text(strip=True)
                company = company_tag.get_text(strip=True)
                loc = location_tag.get_text(strip=True) if location_tag else "Belirtilmemiş"
                link = link_tag.get("href", "").split("?")[0]
                job_id = link.rstrip("/").split("-")[-1]

                job_list.append(JobListing(
                    title=title,
                    company=company,
                    location=loc,
                    job_link=link,
                    job_id=job_id
                ))
        return job_list
    except Exception as e:
        print(f"[-] İlan listesi çekilirken hata: {e}")
        return []

def fetch_job_details(job_id: str) -> str:
    """İlanın detay sayfasına giderek iş tanımı metnini çeker."""
    detail_url = f"https://www.linkedin.com/jobs-guest/jobs/api/jobPosting/{job_id}"
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "Accept-Language": "tr-TR,tr;q=0.9,en-US;q=0.8,en;q=0.7"
    }
    try:
        response = requests.get(detail_url, headers=headers, timeout=10)
        if response.status_code != 200:
            return "Detay metni çekilemedi."

        soup = BeautifulSoup(response.text, "html.parser")
        description_div = soup.find("div", class_="show-more-less-html__markup")
        return description_div.get_text(separator="\n", strip=True) if description_div else "Açıklama bulunamadı."
    except Exception as e:
        return f"Hata: {str(e)}"

# ==========================================
# 4. GEMINI İLE PUANLAMA FONKSİYONU (FAZ 2)
# ==========================================
def evaluate_job_with_gemini(job_title: str, company: str, raw_description: str) -> JobMatchAnalysis:
    """İlanı Gemini ile analiz eder ve adayın profiline göre puanlar."""
    candidate_profile = """
    ADAY PROFİLİ:
    - Eğitim: Endüstri Mühendisliği Lisans Mezunu.
    - Seviye: Junior / Uzman Yardımcısı / Giriş Seviyesi.
    - Yetkinlik & İlgi Alanları:
      1. Veri Analitiği (Python, SQL, BI araçları, veri modelleme ve karar destek sistemleri).
      2. Tedarik Zinciri & Lojistik, Üretim Planlama & Kontrol, Süreç Optimizasyonu.
      3. Proje Yönetimi, İş Analizi ve Pazarlama Analitiği.
    - Hedef: Mühendislik formasyonunu veri analitiği, tedarik zinciri veya üretim/süreç yönetimi rollerinde kullanmak.
    """

    prompt = f"""
    Sen uzman bir İnsan Kaynakları ve Teknik Kariyer Danışmanısın.

    Aşağıda aday profili ve bir iş ilanının ham metni verilmiştir.

    GÖREVİN:
    1. İlan metnindeki kurumsal tanıtım, vizyon/misyon gibi dolgu metinleri tamamen yok say.
    2. Doğrudan ilanın aradığı ana niteliklere, teknik araçlara (SQL, Python, Excel, SAP vb.), tecrübe yılına ve sorumluluklara odaklan.
    3. Adayın mezuniyet ve uzman yardımcısı/junior seviyesini göz önünde bulundurarak ilanı 0-100 arasında objektif olarak puanla.
    4. Senior (5+ yıl) veya alakasız pozisyonlar için düşük puan; Junior Veri Analisti, Tedarik Zinciri, Üretim Planlama gibi roller için yüksek puan ver.

    {candidate_profile}

    ---
    İLAN BİLGİLERİ:
    Pozisyon: {job_title}
    Şirket: {company}

    HAM İLAN METNİ:
    {raw_description}
    """

    response = client.models.generate_content(
        model='gemini-flash-latest',
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=JobMatchAnalysis,
            temperature=0.2,
        ),
    )

    return JobMatchAnalysis.model_validate_json(response.text)

# ==========================================
# 5. TEST VE ÇALIŞTIRMA
# ==========================================
import time

# 1. Arama Yap ve İlanları Çek (Tek Seferde)
keyword_to_search = "Data Analyst"
print(f"🔍 '{keyword_to_search}' pozisyonu için ilanlar çekiliyor...")

# Test için ilk 3 ilanı çekelim
job_results = fetch_linkedin_jobs(keyword=keyword_to_search, location="Istanbul, Turkey", limit=3)

print(f"\n✅ Toplam {len(job_results)} adet ilan bulundu. Gemini analizleri başlıyor...\n")

analyzed_jobs = []

# 2. Çekilen İlanları Sırayla Döngüye Al ve Analiz Et
for idx, job in enumerate(job_results, 1):
    print(f"[{idx}/{len(job_results)}] Analiz Ediliyor: {job.title} - {job.company}...")

    # Detay metnini çek
    job_desc = fetch_job_details(job.job_id)

    # Gemini ile puanla
    analysis = evaluate_job_with_gemini(
        job_title=job.title,
        company=job.company,
        raw_description=job_desc
    )

    # Sonucu sakla
    analyzed_jobs.append({
        "job": job,
        "analysis": analysis
    })

    # API limitlerine takılmamak için kısa bekleme
    time.sleep(1)

# 3. Sonuç Raporu (Puan Sırasına Göre Diz)
print("\n" + "="*60)
print("📊 GÜNCEL İLAN ANALİZ VE PUANLAMA RAPORU")
print("="*60)

# Puanı yüksekten düşüğe sırala
analyzed_jobs.sort(key=lambda x: x["analysis"].match_score, reverse=True)

for item in analyzed_jobs:
    j = item["job"]
    a = item["analysis"]

    print(f"\n🎯 [{a.match_score}/100] - {j.title} | {j.company}")
    print(f"📍 Konum: {j.location}")
    print(f"🔗 Link: {j.job_link}")
    print(f"📌 Durum: {a.suitability_category}")
    print(f"📝 Danışman Notu: {a.brief_summary}")
    print(f"✅ Artılar: {', '.join(a.matching_points[:2])}")
    print(f"⚠️ Riskler: {', '.join(a.missing_or_risk_points[:2])}")
    print("-" * 50)
```

```python
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from google.colab import userdata

# 1. E-Posta Kimlik Bilgilerini Oku
try:
    SENDER_EMAIL = userdata.get('SENDER_EMAIL')
    EMAIL_PASSWORD = userdata.get('EMAIL_PASSWORD')
    RECEIVER_EMAIL = SENDER_EMAIL  # Raporu kendi e-postana göndereceğiz
except Exception:
    # Secrets tanımlı değilse test için doğrudan buraya yazabilirsin:
    SENDER_EMAIL = "SENIN_GMAIL_ADRESIN@gmail.com"
    EMAIL_PASSWORD = "16_HANELI_UYGULAMA_SIFRESI"
    RECEIVER_EMAIL = SENDER_EMAIL


# 2. HTML Formatında Mail Gövdesi Hazırlama
def generate_email_html(high_match_jobs: list) -> str:
    """
    Yüksek puanlı ilanları ve Gemini analizlerini şık bir HTML bültenine dönüştürür.
    """
    html_content = f"""
    <!DOCTYPE html>
    <html>
    <head>
      <meta charset="utf-8">
      <style>
        body {{ font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; background-color: #f4f6f9; margin: 0; padding: 20px; }}
        .container {{ max-width: 650px; margin: 0 auto; background: #ffffff; border-radius: 8px; padding: 25px; box-shadow: 0 2px 5px rgba(0,0,0,0.1); }}
        .header {{ border-bottom: 2px solid #2563eb; padding-bottom: 12px; margin-bottom: 20px; }}
        .header h2 {{ color: #1e293b; margin: 0; }}
        .job-card {{ border: 1px solid #e2e8f0; border-radius: 6px; padding: 18px; margin-bottom: 18px; background: #fafafa; }}
        .job-title {{ font-size: 17px; font-weight: bold; color: #1e40af; margin-bottom: 4px; }}
        .job-meta {{ font-size: 13px; color: #64748b; margin-bottom: 10px; }}
        .badge {{ display: inline-block; padding: 4px 10px; font-size: 12px; font-weight: bold; border-radius: 12px; background-color: #10b981; color: white; }}
        .summary {{ font-size: 14px; color: #334155; margin: 10px 0; line-height: 1.4; }}
        .list-section {{ font-size: 13px; margin: 8px 0; }}
        .list-section strong {{ color: #0f172a; }}
        .btn {{ display: inline-block; margin-top: 10px; padding: 8px 14px; font-size: 13px; font-weight: bold; color: #ffffff !important; background-color: #2563eb; text-decoration: none; border-radius: 4px; }}
        .footer {{ font-size: 12px; color: #94a3b8; text-align: center; margin-top: 25px; border-top: 1px solid #e2e8f0; padding-top: 10px; }}
      </style>
    </head>
    <body>
      <div class="container">
        <div class="header">
          <h2>🎯 Günlük Kişiselleştirilmiş İş İlanı Bülteni</h2>
          <p style="color: #64748b; font-size: 14px; margin-top: 5px;">Profilinize en uygun fırsatlar Gemini AI tarafından filtrelendi.</p>
        </div>
    """

    for item in high_match_jobs:
        job = item["job"]
        analysis = item["analysis"]

        html_content += f"""
        <div class="job-card">
          <div style="display: flex; justify-content: space-between; align-items: center;">
            <div class="job-title">{job.title}</div>
            <span class="badge">Uyum: {analysis.match_score}/100</span>
          </div>
          <div class="job-meta">🏢 {job.company} | 📍 {job.location}</div>
          <div class="summary"><strong>📝 AI Değerlendirmesi:</strong> {analysis.brief_summary}</div>

          <div class="list-section">
            <strong>✅ Eşleşen Yönler:</strong> {", ".join(analysis.matching_points[:2])}
          </div>
          <div class="list-section">
            <strong>⚠️ Risk / Dikkat Noktaları:</strong> {", ".join(analysis.missing_or_risk_points[:2])}
          </div>

          <a href="{job.job_link}" class="btn" target="_blank">İlana Başvur / İncele →</a>
        </div>
        """

    html_content += """
        <div class="footer">
          Bu e-posta, Google Colab & Gemini AI Otomasyon Botu tarafından otomatik oluşturulmuştur.
        </div>
      </div>
    </body>
    </html>
    """
    return html_content


# 3. SMTP ile E-Posta Gönderme Fonksiyonu
def send_email_report(subject: str, html_body: str) -> bool:
    """
    Gmail SMTP sunucusu üzerinden e-posta gönderir.
    """
    try:
        msg = MIMEMultipart("alternative")
        msg["Subject"] = subject
        msg["From"] = f"İş İlanı Takip Botu <{SENDER_EMAIL}>"
        msg["To"] = RECEIVER_EMAIL

        # HTML içeriğini ekle
        msg.attach(MIMEText(html_body, "html", "utf-8"))

        # Gmail SMTP Bağlantısı (SSL - Port 465)
        with smtplib.SMTP_SSL("smtp.gmail.com", 465) as server:
            server.login(SENDER_EMAIL, EMAIL_PASSWORD)
            server.sendmail(SENDER_EMAIL, RECEIVER_EMAIL, msg.as_string())

        print(f"📧 [BAŞARILI] E-posta başarıyla gönderildi: {RECEIVER_EMAIL}")
        return True

    except Exception as e:
        print(f"[-] [HATA] E-posta gönderilemedi: {str(e)}")
        return False
```

```python
# Analiz edilen ilanlardan eşik puanı (örneğin 60 veya 70 puan) geçenleri filtrele
score_threshold = 60
filtered_jobs = [item for item in analyzed_jobs if item["analysis"].match_score >= score_threshold]

if filtered_jobs:
    print(f"🎯 Eşik puanı ({score_threshold}+) geçen {len(filtered_jobs)} ilan bulundu. Mail hazırlanıyor...")
    email_html = generate_email_html(filtered_jobs)

    # Maili gönder
    send_email_report(
        subject=f"🚀 Günün Eşleşen İş İlanları ({len(filtered_jobs)} Yeni Fırsat)",
        html_body=email_html
    )
else:
    print(f"[-] {score_threshold} eşik puanını geçen ilan bulunamadı.")
```
